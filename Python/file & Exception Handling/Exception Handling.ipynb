#### Understanding Exceptions

Exception Handling in Python Allows you to handle errors gracefully and take corrective actions without Stoping the execution of the program. try,except,else,and finally blocks 


#### what is Exceptions?
Exception are events that disrupt the Normal flow of a program. They occur when an error is encountered during program execution.
##### Common Exceptions include:
- ZeroDivisionError : Dividing by Zero
- FileNotFoundEror : File not Found
- ValueError - Invalid value
- TypeError - Invalid type

```python
try:
    num = int(input('Enter a Number'))
    result = 10/num
    print(result)
    
except ValueError:
    print('This is not a valid number')

except ZeroDivisionError:
    print("Enter denominator greater than 0")

except Exception as ex:
    print(ex)
```

```python
# try,except,else,block 
try:
    num = int(input('Enter a Number'))
    result = 10/num
except ValueError:
    print('This is not a valid number')

except ZeroDivisionError:
    print("Enter denominator greater than 0")
    
except Exception as ex:
    print(ex)
    
else:
    print(f'The result is {result}')
    
```

```python
## try , except , else, and finally 
try:
    num = int(input('Enter a Number'))
    result = 10/num
except ValueError:
    print('This is not a valid number')
except ZeroDivisionError:
    print("Enter denominator greater than 0")  
except Exception as ex:
    print(ex) 
else:
    print(f'The result is {result}')
finally:  # Weather any block of code is running or not it must be executed
    print('Exception Complete')
```
