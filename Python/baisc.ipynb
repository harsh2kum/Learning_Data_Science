```python
def fact(n):
    if n == 0:
        return 1
    else:
        return n*fact(n-1)
    
#function call 
num = int(input("enter the number:"))
print(fact(num))
```

### if-else, Loops


```python
# Nested if 
number = float(input("Enter a number: "))
if number > 0:
    if number % 2 == 0:
        print("The number is positive and even.")
    else:
        print("The number is positive and odd.")
else:
    print("The number is negative.")
```

```python
#for loop
for i in range(11):
    print(i)
```

```python
#wile loop
i = 1
while i<=10:
    print(i)
    i += 1
```

```python
# 5x5 '*' using nestedloop
for i in range(5):
    for j in range(5):
        print('*',end = '')
    print() 
```

```python
for i in range(1,11):
    if i == 5:
        continue
    print(i)
    
```
