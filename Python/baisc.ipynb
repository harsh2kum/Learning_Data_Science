```python
# Guessing Game
import random
jackpot = random.randint(1,100)

guess_number = int(input('enter your number: '))

count = 1
while guess_number != jackpot:
    
    if guess_number < jackpot:
        print('guess higher number')
    else:
        print('Wrong,guess lower number')
    guess_number = int(input('enter your guess number'))
    count += 1
    
else:
    print('Correct Guess')
    print('your guess count',count)
```
