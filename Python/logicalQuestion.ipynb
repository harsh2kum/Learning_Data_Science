Phase 1 – Conditional Thinking (If–Else, Boolean Logic)

```python
# 1. Take a number and print whether it’s positive, negative, or zero.
num = int(input("enter the number: "))
if num == 0:
    print('Zero')
elif num > 0 :
    print("Positive")
else:
    print("negative")
```

```python
# 2. Check if a number is even or odd
num = int(input('enter the number:'))
if num%2==0:
    print("Even")
else:
    print('Odd')
# or 
print('Even' if num%2 == 0 else 'Odd')
```

```python
# 3.Check if a number is divisible by 5.
num = int(input("Enter the number: "))
if num%5==0:
    print('divisible by 5 ')
else:
    print("not divisible by 5")
# or 
print("Divisible by 5" if num % 5 == 0 else "Not divisible by 5")
```

```python
# 4. Check if a number is divisible by both 3 and 5.
num = int(input("Enter the number: "))
if num%3 == 0 and num%5==0:
    print("number is divisible by both 3 and 5.")
else:
    print('Not Divisible by Both')
```

```python
# 5. Check if a given year is a leap year.
year = int(input('enter the year:'))
if (year%4 == 0 and year%100 != 0) or (year%400==0):
    print('Leap Year')
else:
    print('Not a leap year')
```

```python
# 6. Take two numbers and print the larger one.
a = int(input("enter 1st number:"))
b = int(input('enter 2nd number '))
if a==b:
    print(f'{a}={b} equal')
elif a>b:
    print(f'{a} is greater')
else:
    print(f'{b} is greater')
    
# or 
print("Larger number is:", a if a > b else b)
```

```python
# 7. Take three numbers and print the largest.
a,b,c = map(int,input('enter the number: ').split())
print('Largest:',max(a,b,c))
```
