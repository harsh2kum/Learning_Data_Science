```python
#  OOP Features in ML App Design

# 1. Using Dataclasses for Input
# Instead of passing raw dictionaries, we can define structured input using dataclass.This makes your code safer and more readable.

from dataclasses import dataclass

@dataclass
class UserInput:
    age:int
    income:float
    education:str
    
raw = UserInput(age=25, income=50000, education="Graduate")

#  2. Adding a Decorator for Logging
# We can log every prediction request automatically using a decorator.

def log_prediction(func):
    def wrapper(*args, **kwargs):
        print("Logging: Prediction requested...")
        result = func(*args, **kwargs)
        print(f"Logging: Prediction result = {result}")
        return result
    return wrapper

class Predictor:
    def __init__(self, model_path: str):
        self.preprocessor = Preprocessor()
        self.model = Model(model_path)

    @log_prediction
    def run(self, raw_input: dict):
        clean_data = self.preprocessor.clean(raw_input)
        prediction = self.model.predict(clean_data)
        return prediction[0]

# 3. Adding Static & Class Methods
# - Static methods → utility functions (e.g., scaling data).
# - Class methods → track usage across all instances. 
class Preprocessor:
    usage_count = 0
    
    def __init__(self):
        Preprocessor.usage_count += 1
    
    @staticmethod
    def normalize(value, min_val, max_val):
        return (value - min_val) / (max_val - min_val)
    
    @classmethod
    def get_usage(cls):
        return cls.usage_count

# 4. Operator Overloading for ML Vectors
# You can make custom classes behave like mathematical objects.

class FeatureVector:
    def __init__(self, values):
        self.values = values
    
    def __add__(self, other):
        return FeatureVector([a+b for a, b in zip(self.values, other.values)])
    
    def __str__(self):
        return f"Vector({self.values})"

v1 = FeatureVector([1,2,3])
v2 = FeatureVector([4,5,6])
print(v1 + v2)   # Vector([5,7,9])

#  5. Generators for Batch Prediction
# Instead of predicting all at once, stream results
class BatchPredictor(Predictor):
    def run_batch(self, inputs: list):
        for raw in inputs:
            yield self.run(raw)
```
