### Iterators
Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1,2,3,4,5,6]
for i in my_list:
    print(i)
```

```python
type(my_list)
print(my_list)
```

```python
## Iterator 
iterator = iter(my_list)
print(type(iterator))
```

```python
iterator
```

```python
## Iterate through all the elements 

next(iterator)
```

```python
iterator = iter(my_list)
```

```python
try:
    print(next(iterator))

except StopIteration:
    print("There are no elements in the iterator")
```

```python
## String iterator 
my_string  = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator))
print(next(string_iterator))
```

### Generators

Generators are a simpler way to create iterators. They use the yield keywords to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(3):
        yield i**2
```

```python
# function call 
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
a
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
    
```

```python
gen = my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

Practical Example : Reading Large Files

Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.



```python
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line 
```

```python
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```

### Decorators

Decorators are a powerful and flexible feature in python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code.

```python
## Function copy - copy 1 function to the other functions
def welcome():
    return "Welcome to the Advance python course"

welcome()
```

```python
wel = welcome
print(wel())
```

```python
del welcome
```

```python
print(wel())
```

```python
## Closures
"""
Closures in Python are inner functions that retain access to variables from their enclosing (outer) function's scope, even after the outer function has finished executing.  This allows the inner function to "remember" and use those variables, effectively creating a function with an extended scope. 
"""
def outer_function(x):
    def inner_function(y):
        return x + y  # 'x' is captured from outer scope
    return inner_function

# Create a closure with x = 10
add_ten = outer_function(10)
print(add_ten(5))  # Output: 15
```

```python
# Closures functions

def main_welcome():
    msg = "Welcome"
    def sub_welcome_method():
        print("Welcome to the Data Science")
        print(msg)
        print("Please learn these concepts properly")
        
    return sub_welcome_method()
```

```python
main_welcome()
```

```python
## Decorator 

def main_welcome(func):

    def sub_welcome_method():
        print("Welcome to the Data Science")
        func()
        print("Please learn these concepts properly")
        
    return sub_welcome_method()
```

```python
def coure_introduction():
    print("This ia an advanced python course")
    
coure_introduction()
```

```python
main_welcome(coure_introduction)
```

```python
@main_welcome
def coure_introduction():
    print("This ia an advanced python course")
```

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening before the function is called.")
        
    return wrapper
```

```python
@my_decorator
def say_hello():
    print('hello')
    
```

```python
say_hello()
```

```python
## Decorator with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print('hello')
    
```

```python
say_hello()
```
