```python
# Data Serialization 
import json
data = {'name':'Harsh','age':25,'Stream':'Data Science'}

json_str = json.dumps(data) #convert dictionary to string
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str) # convert string to dictionary 
print(parsed_data)
print(type(parsed_data))
```

```python
# CSV
import csv

with open('example.csv',mode='w',newline='') as file:
    write = csv.writer(file)
    write.writerow(['name','age']) #row1 
    write.writerow(['Harsh',25]) #row2
    
#open and read that file    
with open('example.csv',mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```
