### Importing Modules and packages 
- In python modules and packages help organize and reuse code.

```python
import math 
math.sqrt(4)
```

```python
# from package_name import package
from math import sqrt,pi,pow
print(sqrt(16))
print(sqrt(49))
print(pow(2,4))
print(pi)
```

```python
from math import *   ##(*) import all the function inside that module 
sqrt(81)
pow(3,2)
```

```python
# from package folder import modules(created by me) # it should be in same folder location 
from package.module import multiply

multiply(4,5)
multiply(9,8)
```

```python
# we can write like this also
from package import module
module.multiply(3,4)

module.subtraction(8,4)
```
