```python
# secret_number  = 7 
# guess = int(input("Guess the number: "))
# attempt = 5
# while attempt > 0:
#     if guess == secret_number:
#         print("You Won")
#         break
#     else:
#         print('wrong guess')
#         attempt -= 1

# correct Answer 

secret_number = 7 

attempts = 5

while attempts > 0:
    guess = int(input('Guess the number: '))
    
    if guess == secret_number:
        print("You Won")
        break
    else:
        attempts -= 1
        print(f'Wrong guess. Attempts left: {attempts}')
        
# Important Programming Lesson
# Input that changes each loop iteration should usually be INSIDE the loop.
# This is a major programming concept.
```

```python
# ❌ Wrong way (input outside the loop)
# Input taken once before loop
n = int(input("Enter the number: "))

for i in range(4):
    print(f'Iteration {i+1}: You entered {n}')
    # Here, the input is fixed. Every iteration just repeats the same value.
```

```python
# ✅ Correct way (input inside the loop)
for i in range(3):
    n = int(input(f'Iteration {i+1} - Enter a number: '))
    print(f'You entered {n}')
    # Each loop iteration asks for new input. The program becomes dynamic because the user can provide different values each time.
```

#### 🔑 Lesson
- Placing input inside the loop ensures that the program can react to changing values, making it interactive and flexible. 
- If the input is outside, the loop just reuses the same value, which often defeats the purpose of looping.

```python
# Simple Calculator 
while True:
    # Take input inside the loop
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operator = input("Enter Operator (+,-,*,/): ")
    
    if operator == "+":
        print("result:", num1 + num2)
    
    elif operator == "-":
        print("Result:", num1 - num2)
        
    elif operator == "*":
        print("Result:", num1 * num2)
        
    elif operator == "/":
        print("Result:", num1/num2)
        
    else:
        print("Invalid Operator!")
        
    # Ask if user want to continue
    choice = input("Do you want another calculation? (yes/no): ")
    if choice.lower() != "yes":
        break
    
"""
🔑 Why input inside the loop matters
- Each time the loop runs, the user can enter new numbers and operators.
- If the input were outside the loop, the calculator would just repeat the same calculation over and over — not useful at all.
- This design makes the program interactive and flexible, which is the whole point of looping with input.
"""
```

```python
# Food Ordering System 

while True:
    print("\n--- Menu ---")
    print("1. Burger")
    print("2. Pizza")
    print("3. Pasta")
    print("4. Exit")
    
    choice = input("Enter your choice (1-4): ")
    
    if choice == "1":
        print("You ordered a Burger")
        
    elif choice == "2":
        print("You Ordered a Pizza")
        
    elif choice == "3":
        print("You ordered Pasta")
    
    elif choice == "4":
        print("Thankyou for Visiting! Goodbye")
        break
    
    else:
        print("Invalid choice, please try again.")
                
```
