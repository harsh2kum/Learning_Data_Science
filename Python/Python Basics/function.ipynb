```python
def greet(name):
    """Greet a person by name"""
    print(f"Starting the greeting function for {name}")
    result = f"Hello, {name}!"
    print("Finishing the greeting function")
    return result

print("Program starts")
message = greet("Harsh")
print(f"Returned message: {message}")
print("Program ends")

''' 
# Here's what happens during function execution:

1.When a function is called, program execution jumps to the function body
2.The function's parameters receive the values provided in the function call (arguments)
3.The function body executes line by line
4.If a return statement is encountered, the function returns the specified value and control goes back to where the function was called
5.If no return statement is encountered, the function returns None and control goes back to where the function was called

'''
```

### `*args and **kwargs`
`*args and **kwargs` are special Python keywords that are used to pass the variable length of arguments to a function

```python
# 4. Variable - Length Arguments - Python allows functions to accept an arbitrary number of arguments
#- *args (Arbitrary Positional Arguments)

def make_pizza(*toppings):
    print("Making a pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza("pepperoni")
make_pizza("mushrooms", "green peppers", "extra cheese")
```

```python
#- *Kwargs(Arbitrary keyword Arguments)
def build_profile(first, last, **user_info):
    """Build a dictionary containing user information."""
    profile = {
        'first_name': first,
        'last_name': last
    }
    
    for key, value in user_info.items():
        profile[key] = value
    
    return profile

user = build_profile('Albert', 'Einstein',
                     location='Princeton',
                     field='Physics',
                     born=1879)
print(user)
```

```python
# Return Statement - The return statement is used to exit a function and pass back a value to the caller. 
# If no return statement is provided, the function returns None by default.
def square(number):
    return number ** 2

result = square(5)  # result = 25
print(result)
```

```python
# Multiple Return Values
# A function can return multiple values by separating them with commas.
# Python actually returns a single tuple containing these values:
def get_dimensions():
    """Return width and height."""
    width = 500
    height = 300
    return width, height

dimensions = get_dimensions()  # dimensions is a tuple (500, 300)
width, height = get_dimensions()  # Unpacking the tuple
print(dimensions)
print(width, height)
```

```python
# Early Return
# A return statement immediately exits the function 
# which can be useful for handling special cases:
def divide(a, b):
    """Divide a by b."""
    if b == 0:
        return "Cannot divide by zero"
    return a // b
divide(12,2)
```

```python
# Returning Different Types
# Python allows functions to return different types based on conditions:
def process_number(num):
    if num > 0:
        return 'Positive' #string
    elif num < 0:
        return -num  # Number (absolute value)
    
    else:
        return None #None
    
print(process_number(-8))
```

```python
#reassign function - we can reassign function in any variable 
y = process_number(10)
print(y)
```

```python
# Variable Scope in Functions
# Variable scope refers to where in your code a variable can be accessed. 
# Python has primarily two scopes:
'''
1.Local Scope
Variables defined inside a function are in local scope and can only be accessed within that function:
'''
def my_function():
    x = 10  # Local variable
    print(x)

my_function()  # Prints 10
print(x)  # Error: x is not defined
```

```python
# Global Scope
# Variables defined outside all functions are in global scope and can be accessed anywhere:

x = 10  # Global variable
def my_function():
    print(x)  # Can access global variable

my_function()  # Prints 10
```

```python
# Modifying Global Variables
# To modify a global variable inside a function, use the global keyword:
x = 10
def modify_global():
    global x
    x = 20  # Modifies the global variable
print(x)  # 10
modify_global()
print(x)  # 20
```

```python
#modifying global variable 
x = 10
def add(): #we have not define any parameter
    global x
    x = 15
    
print(x)
add() # we can call it directly
print(x)
```

#### Lambda function 
- Lambda function can ktake any number of arguments, but can only have one expression.
- lambda has no return value(infact it return a function)
- lambda is written in 1 line 
- not resuable 
- lambda function is use for Higher order function (HOF)

#### Diff between lambda vs Normal Function

- No name
- lambda has no return value(infact,returns a function)
- lambda is written in 1 line
- not reusable

Then why use lambda functions?<br>
**They are used with HOF**


```python
# Parameter - one of more parameters are supported must be separated by a comma(,) and no parentheses--eg: a,b
# colon(:) ois a cue for the expression 
# lambda keyword - Creates the lambda expression
#Expression - Must be a single valid python expression -eg - a + b
lambda a,b: a+b
```

```python
a = lambda x:x**2
a(4)
```

```python
a = lambda x,y:x+y
a(4,5)
```

```python
# check if a string has 'a'
a = lambda s:'a' in s
a('hello')
a('air')

#odd or even 
a = lambda x:'even' if x%2==0 else 'odd'
a(6)
a(9)

#Higher Order function - if a function take another function as input or return function it is call higher order function 
L = [1,2,3,4,5]
def square(x):
    return x**2

def Transform(f,L): #HOF
    output = []
    for i in L:
        output.append(f(i))
    print(output)
    
#function call 
Transform(square,L)
Transform(lambda x:x**2,L)
```

### Higher Order Function (map,filter and Reduce)

#### map

```python
#square the items of a list
list(map(lambda x:x**2,[1,2,3,4,5]))

#odd/even labelling of list items
L = [1,2,3,4,5,6]
list(map(lambda x:'even' if x%2==0 else 'odd',L))
```

```python
#fetch names from a list of dict 
users = [
    {
        'name': "Harsh",
        'age': 25,
        'gender':'male'
    },
    
    {
        'name':'Bappy',
        'age':27,
        'gender':'male'
    }
    ,
    {
        'name':'monisha',
        'age': 22,
        'gender': 'Female'
            
    }
]

list(map(lambda users:users['name'],users))
list(map(lambda users:users['age'],users))
list(map(lambda users:users['gender'],users))
```

#### Filter

```python
# Number greater than 2
L = [1,2,3,4,5]
list(filter(lambda x:x>2,L))
```

```python
# Fetch fruits starting with letter 'a'
fruits = ['apple','guava','cherry']
list(filter(lambda x:x.startswith('a'),fruits))
```

### Reduce
- There is no reduce function we have to import library - import functools 


```python
# sum of all item
import functools
L = [1,2,3,4,5]
functools.reduce(lambda x,y:x+y,L)
```

```python
# find maximum number form the list 
functools.reduce(lambda x,y:x if x>y else y,[23,11,25,9,30,3])
# find minimum number form the list 
functools.reduce(lambda x,y:x if x<y else y,[23,11,25,9,30,3])
```
