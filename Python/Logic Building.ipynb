```python
# 1. Take three sides and check if they form a valid triangle
# 2. If the sides form a valid triangle, determine whether it is equilateral, isosceles, or scalene.

# Valid Triangle -  Sum of any two sides must be greater than the third side.
# Equilateral: all sides equal
# Isosceles: any two sides equal
# Scalene: all sides different 

# user input 
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

#check  a,b,c are positive 
if a>0 and b>0 and c>0:
    # check triangle condition
    if (a + b > c) and (b + c > a) and (a + c > b):
        print("Valid Triangle")
        
        # check type of triangle
        if a==b and b == c:
            print("Triangle is Equilateral")
        elif a==b or b==c or a==c:
            print("Triangle is Isosceles")
        else:
            print("Triangle is Scalene")
            
    else:
        print("Not a Valid Triangle")
else:
    print("Sides must be positive number")
```

### Level 3: Math and Number Logic

```python
# 1. Take a 3-digit number and check if all digits are distinct
# Take input from user
num = int(input("Enter a 3-digit number: "))

# Convert to positive (handle negative numbers)
if num < 0:
    num = -num

# Check if 3-digit number
if num >= 100 and num <= 999:

    # Extract digits
    a = num // 100 #472 // 100 = 4
    b = (num // 10) % 10 #(472 // 10) % 10 = 47 % 10 = 7
    c = num % 10 # 472 % 10 = 2

    # Nested if to check distinct digits
    if a != b:
        if b != c:
            if a != c:
                print("All digits are distinct")
            else:
                print("Digits are not distinct")
        else:
            print("Digits are not distinct")
    else:
        print("Digits are not distinct")

else:
    print("Please enter a valid 3-digit number")
```

```python
def check_distinct_digits(num):
    # Convert to positive (handle negative numbers)
    if num < 0:
        num = -num

    # Check if 3-digit number
    if 100 <= num <= 999:
        # Extract digits
        a = num // 100          # hundreds place
        b = (num // 10) % 10    # tens place
        c = num % 10            # ones place

        # Check distinct digits
        if a != b and b != c and a != c:
            print("All digits are distinct")
        else:
            print("Digits are not distinct")
    else:
        print("Please enter a valid 3-digit number")


# Example usage
num = int(input("Enter a 3-digit number: "))
check_distinct_digits(num)
```

```python
# Function 
def extract_digits(num):
    a = num//100 # hundreds place
    b = (num//10)%10  # tens place
    c = num%10 #ones place
    return a,b,c
#function call 
num = int(input("Enter a 3-digit number: "))
hundreds, tens, ones = extract_digits(num)
print("Hundreds digit:", hundreds)
print("Tens digit:", tens)
print("Ones digit:", ones)
```

```python
def extract_digits_and_check(num):
    # Extract digits (works for any length number)
    digits = []
    while num > 0:
        digits.append(num % 10)   # take last digit
        num //= 10                # remove last digit
    
    digits = digits[::-1]         # reverse to correct order
    
    # Check if all digits are distinct
    if len(digits) == len(set(digits)):
        print("Digits:", digits)
        print("All digits are distinct ")
    else:
        print("Digits:", digits)
        print("Digits are NOT distinct ")

# Example usage
num = int(input("Enter a number: "))
extract_digits_and_check(num)
```

```python
# - For numbers → you can swap directly with a, b = b, a.
# - For characters in a string → you need to convert to a list first, swap, then join back
def swap_numbers(a, b):
    print("Before swap: a =", a, "b =", b)
    a, b = b, a   # Python’s tuple unpacking makes swapping easy
    print("After swap: a =", a, "b =", b)

# Example usage
swap_numbers(10, 20)

def swap_chars(s, i, j):
    # Convert string to list (since strings are immutable)
    chars = list(s)
    chars[i], chars[j] = chars[j], chars[i]  # swap
    return "".join(chars)

# Example usage
word = "hello"
new_word = swap_chars(word, 0, 4)  # swap 'h' and 'o'
print("Before swap:", word)
print("After swap:", new_word)
```
