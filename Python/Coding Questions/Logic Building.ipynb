```python
# Recursion
# A function can call itself — useful for problems like factorial, Fibonacci, or tree traversal.

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
print(factorial(7))
```

```python
# Higher-Order Functions
# Functions can take other functions as arguments or return them.

def apply_twice(func,value):
    return func(func(value))

def increment(x):
    return x + 1

print(apply_twice(increment,5))
print(apply_twice(increment,9))
```

```python
# Decorators
# Special functions that modify other functions.

def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print('After function')
    return wrapper

@decorator
def say_hello():
    print('Hello!')

say_hello()
```

```python
#  Generators (yield)
# Functions that produce values one at a time instead of returning everything at once.

def countdown(n):
    while n > 0:
        yield n
        n -= 1
        
for num in countdown(5):
    print(num)
```

```python
# Encapsulation
# - Hides internal details and exposes only necessary parts.
# - Achieved using private attributes (_ or __).

# Encapsulation - Hide details - Private balance in BankAccount
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance  # Private Attribute
        
    def deposit(self,amount):
        self.__balance += amount
        
    def get_balance(self):
        return self.__balance 
    
    def withdraw(self):
        self.__balance -= amount
        
    def withdraw(self, amount):  
        if amount > 0 and amount <= self.__balance:  
            self.__balance -= amount
            return f"Withdrew ${amount}"  
        else:
            return "Insufficient funds"
    
acc = BankAccount(10000)
acc.deposit(500)

print('Balance',acc.get_balance())
print("Withdraw",acc.withdraw(1000))
```

```python
# Inheritance
# - Allows one class to inherit properties/methods from another.
# - Promotes reusability.
# - Reuse code
class Animal:
    def speak(self):
        print('I am an animal')

# Dog inherits from Animal      
class Dog(Animal):
    def speak(self):
        print('Woof!')

class Cat(Animal):
    def speak(self):
        print('Meow!')

d = Dog()
d.speak()
c = Cat()
c.speak()
```

```python
# Polymorphism
# - Same method name, different behavior depending on the object.
# eg - sound() in Bird vs Cat
class Bird:
    def sound(self):
        print("Chirp")
        
class Cat:
    def sound(self):
        print("Meow")
    
for animal in [Bird(),Cat()]:
    animal.sound()
```

```python
# Abstraction
# - Hides implementation details, shows only essential features.
# - Achieved using abstract classes (abc module).

#eg - Abstract Shape class

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self,side):
        self.side = side

    def area(self):
        return self.side * self.side
    
sq = Square(4)
print("Area",sq.area())
```

```python
# Class Methods
# - Operate on the class itself, not just instances.
# - Defined with @classmethod 

class Person:
    count = 0
    
    def __init__(self,name):
        self.name = name
        Person.count += 1
        
    @classmethod 
    def total_people(cls):
        return cls.count
    
p1 = Person("Harsh")
p2 = Person("Amit")
p3 = Person("Rohit")
print(Person.total_people())
```

```python
# Magic / Dunder Methods
# Special methods that start and end with __.
# Examples:
#  __init__ → constructor
#  __str__ → string representation
#  __len__ → length
#  __add__ → operator overloading

class Book:
    def __init__(self,title,pages):
        self.title = title
        self.pages = pages
        
    def __str__(self):
        return f'{self.title} ({self.pages} pages)'
    
    def __len__(self):
        return self.pages
    
b = Book("Python Basics", 250)
print(b)
print(len(b))
```

```python
# Operator Overloading
# You can redefine how operators work for your class

class Vector:
    def __init__(self,x,y):
        self.x, self.y = x,y
        
    def __add__(self,other):
        return Vector(self.x + other.x, self.y + other.y)
    
v1 = Vector(2,3)
v2 = Vector(4,5)
v3 = v1 + v2
print(v3.x, v3.y)
```

```python
# Multiple Inheritance
# A class can inherit from more than one parent.

class A:
    def method_a(self): print("A")
    
class B:
    def method_b(self): print("B")
    
class C(A,B):
    pass
    
c = C()
c.method_a()
c.method_b()
```

```python
# 1. Take a 3-digit number and check if all digits are distinct
# Take input from user
num = int(input("Enter a 3-digit number: "))

# Convert to positive (handle negative numbers)
if num < 0:
    num = -num

# Check if 3-digit number
if num >= 100 and num <= 999:

    # Extract digits
    a = num // 100 #472 // 100 = 4
    b = (num // 10) % 10 #(472 // 10) % 10 = 47 % 10 = 7
    c = num % 10 # 472 % 10 = 2

    # Nested if to check distinct digits
    if a != b:
        if b != c:
            if a != c:
                print("All digits are distinct")
            else:
                print("Digits are not distinct")
        else:
            print("Digits are not distinct")
    else:
        print("Digits are not distinct")

else:
    print("Please enter a valid 3-digit number")
```

```python
def check_distinct_digits(num):
    # Convert to positive (handle negative numbers)
    if num < 0:
        num = -num

    # Check if 3-digit number
    if 100 <= num <= 999:
        # Extract digits
        a = num // 100          # hundreds place
        b = (num // 10) % 10    # tens place
        c = num % 10            # ones place

        # Check distinct digits
        if a != b and b != c and a != c:
            print("All digits are distinct")
        else:
            print("Digits are not distinct")
    else:
        print("Please enter a valid 3-digit number")


# Example usage
num = int(input("Enter a 3-digit number: "))
check_distinct_digits(num)
```

```python
# Function 
def extract_digits(num):
    a = num//100 # hundreds place
    b = (num//10)%10  # tens place
    c = num%10 #ones place
    return a,b,c
#function call 
num = int(input("Enter a 3-digit number: "))
hundreds, tens, ones = extract_digits(num)
print("Hundreds digit:", hundreds)
print("Tens digit:", tens)
print("Ones digit:", ones)
```

```python
def extract_digits_and_check(num):
    # Extract digits (works for any length number)
    digits = []
    while num > 0:
        digits.append(num % 10)   # take last digit
        num //= 10                # remove last digit
    
    digits = digits[::-1]         # reverse to correct order
    
    # Check if all digits are distinct
    if len(digits) == len(set(digits)):
        print("Digits:", digits)
        print("All digits are distinct ")
    else:
        print("Digits:", digits)
        print("Digits are NOT distinct ")

# Example usage
num = int(input("Enter a number: "))
extract_digits_and_check(num)
```

```python
# - For numbers → you can swap directly with a, b = b, a.
# - For characters in a string → you need to convert to a list first, swap, then join back
def swap_numbers(a, b):
    print("Before swap: a =", a, "b =", b)
    a, b = b, a   # Python’s tuple unpacking makes swapping easy
    print("After swap: a =", a, "b =", b)

# Example usage
swap_numbers(10, 20)

def swap_chars(s, i, j):
    # Convert string to list (since strings are immutable)
    chars = list(s)
    chars[i], chars[j] = chars[j], chars[i]  # swap
    return "".join(chars)

# Example usage
word = "hello"
new_word = swap_chars(word, 0, 4)  # swap 'h' and 'o'
print("Before swap:", word)
print("After swap:", new_word)
```

```python
# 2. Take a 3-digit number and determine if the middle digit is the largest, smallest, or neither.
def middle_digit_relation(n):
    if not (100 <= n <= 999):
        raise ValueError('n must be a 3 digit Number')
    h = n // 100
    m = (n // 10) % 10
    u = n % 10
    if m>h and m>u:
        return 'Largest'
    elif m<h and m<u:
        return 'Smallest'
    else:
        return 'Neither'
#Function Call 
a = int(input("Enter 3 digit Number: "))
middle_digit_relation(a)

 
# - Extract digits:
# - Hundreds digit: h=[n/100]
# - Middle digit: m=[n/10] - 10 * h
# - Units digit: u=n-10[n/10] 
# - Compare:
# - If m>h and m>u, then the middle digit is the largest.
# - If m<h and m<u, then the middle digit is the smallest.
# - Otherwise, it’s neither (it could be equal to one or both, or strictly between them).
# Examples
# - n=583: h=5,m=8,u=3 - m is largest.
# - n=747: h=7,m=4,u=7 - m is smallest.
# - n=444: h=4,m=4,u=4 - m is neither.
```

```python
num = int(input("Enter a 3-digit number: "))

# Extract digits
hundreds = num // 100
middle   = (num // 10) % 10
units    = num % 10

# Compare using if-else
if middle > hundreds and middle > units:
    print("Middle digit is the largest")
elif middle < hundreds and middle < units:
    print("Middle digit is the smallest")
else:
    print("Middle digit is neither largest nor smallest")
```

```python
# 3. Take a 4-digit number and check if the first and last digits are equal
num = int(input('Enter the 4-digit number: '))

# Extract digits
first_digit = num // 1000 # integer division gives the thousands place
last_digit = num % 10 # modulo gives the units place

# Compare using if-else
if first_digit == last_digit:
    print("First and last digits are equal")
else:
    print("First and last digits are not equal")

#Extract all 4 digits
# Thousands = num // 1000
#     hundreds  = (num // 100) % 10
#     tens      = (num // 10) % 10
#     ones      = num % 10
```
