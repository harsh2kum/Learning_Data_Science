### Binary Classification Problem

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
%matplotlib inline 
```

```python
from sklearn.datasets import make_classification
```

```python
## Create the dataset
X,y = make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
X
# to convert x into dataframe 
pd.DataFrame(X)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
# Model training 
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression() 
```

```python
logistic.fit(X_train,y_train)
```

```python
#prediction datapoints
y_pred =  logistic.predict(X_test)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
```

```python
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
print(classification_report(y_test,y_pred))
```

### Hyperparameter Tuning and Cross Validation 

```python
model = LogisticRegression()
penalty  = ['l1','l2','elasticnet']
c_values = [100,10,1.0,0.1,0.01]
solver = ['newton-cg','lbfgs','liblinear','sag','saga']
```

```python
params = {
    'penalty': penalty,
    'C': c_values,
    'solver': solver
}
```

```python
params.keys()
params.values()
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
```

```python
## GridSearchCV
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

### Randomized SearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
randomcv = RandomizedSearchCV(estimator=model,param_distributions=params,cv=5,scoring='accuracy')
```

```python
randomcv.fit(X_train,y_train)
```

```python
randomcv.best_score_
```

```python
randomcv.best_params_
```

```python
y_pred = randomcv.predict(X_test)
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

### Logistic Regression for Multiclass Classification Problem 

```python
X,y = make_classification(n_samples=1000,n_features=10,n_informative=3,n_classes=3,random_state=15)
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression(multi_class='ovr') #one vs rest (ovr)
logistic.fit(X_train,y_train)
y_pred = logistic.predict(X_test)
```

```python
y_pred
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

### Logistic Regression for Imbalanced Dataset

```python
# Generate and plot a synthetic imbalanced classification dataset
from collections import Counter
from sklearn.datasets import make_classification
```

```python
## Imbalanced dataset 
X,y = make_classification(n_samples=10000,n_features=2,n_clusters_per_class=1,
                          n_redundant=0,weights=[0.99],random_state=10)
```

```python
X
```

```python
Counter(y)
```

```python
import pandas as pd
import seaborn as sns

# Convert X into a DataFrame with named columns
df = pd.DataFrame(X, columns=["feature1", "feature2"])
df["target"] = y   # add target column

# Scatterplot
sns.scatterplot(x="feature1", y="feature2", hue="target", data=df)
```

```python
# Split the dataset into train and test 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=0)
```

```python
class_weight = [{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100] for y in [1,10,50,100]]
```

```python
class_weight
```

```python
# Hyperparameter tunning 
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
penalty = ['l1','l2','elasticnet']
c_values = [100,10,1.0,0.1,0.01]
solver = ['newton-cg','lbfgs','liblinear','sag','saga']
class_weight = [{0:w,1:y} for w in [1,10,50,100] for y in [1,10,50,100] for y in [1,10,50,100]]
```

```python
params = dict(penalty=penalty,C=c_values,solver=solver,class_weight=class_weight)
```

```python
params
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
grid = GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
score = accuracy_score(y_pred,y_test)
print(score)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```
