```python
import pandas as pd
import numpy as np
```

```python
messages = pd.read_csv("SMSSpamCollection.txt", sep="\t", names=['label','message'])
```

```python
messages
```

```python
# Data Cleaning and Preprocessing
import re # Regular expresion
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

### Create Bag of words

```python
## Output Features
y = pd.get_dummies(messages['label'])
y = y.iloc[:,0].values
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_test,y_test = train_test_split(corpus,y,test_size=0.2,random_state=42)
```

```python
# Create the Bag Of Words Model 
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
len(X_train),len(y)
```

```python
# Independent feature 
X_train = cv.fit_transform(X_train).toarray()
X_test = cv.transform(X_test).toarray()
```

```python
np.set_printoptions(edgeitems=30,linewidth=100000,
                    formatter=dict(float=lambda x: "%.3g" % x))

X_train
```

```python
cv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_detect_model = MultinomialNB()
```

```python
spam_detect_model.fit(X_train,y_train)
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
y_pred
```

```python
# compar y_test & y_pred 
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

### Create The TF-IDF Model 

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_test,y_test = train_test_split(corpus,y,test_size=0.2)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tv = TfidfVectorizer(max_features=2500, ngram_range=(1,2))
X_train = tv.fit_transform(X_train).toarray()
X_test = tv.transform(X_test).toarray()
```

```python
X_train
```

```python
tv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_tfidf_detect_model = MultinomialNB().fit(X_train,y_train)
```

```python
# Prediction 
y_pred = spam_tfidf_detect_model.predict(X_test)
```

```python
y_pred
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
```
