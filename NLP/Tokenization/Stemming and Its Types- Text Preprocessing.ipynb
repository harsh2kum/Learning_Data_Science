## Stemming
Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).

```python
## Classification Problem
## Comments of product is a positive review or negative review
## Reviews----> eating, eat,eaten, [going,gone,goes]--->go

words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
words
```

### Stemming Techniques:

1. Porter Stemmer: A widely used algorithm is introduced, with demonstrations of its application using NLTK.
2. Regular Expression Stemmer: This allows for customized stemming based on specific patterns in words.
3. Snowball Stemmer: Presented as an improvement over the Porter Stemmer, it offers better accuracy and handles multiple languages effectively.

Limitations of Stemming: The drawbacks of stemming that potentially losing the original meaning of words, which can lead to inaccuracies.

### PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem("sitting")
```

### RegexpStemmer class
NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression.

A stemmer that uses regular expressions to identify morphological affixes. Any substrings that match the regular expressions will be removed.

```python
from nltk.stem import RegexpStemmer
```

```python
st = RegexpStemmer('ing$|s$|e$|able$', min=4)
print(st.stem('cars'))
print(st.stem('mass'))
print(st.stem('doing'))
print(st.stem('does'))
# remove s,ing,e,able
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

### Snowball Stemmer
 It is a stemming algorithm which is also known as the Porter2 stemming algorithm as it is a better version of the Porter Stemmer since some issues of it were fixed in this stemmer.

```python
from nltk.stem import SnowballStemmer
```

```python
snowballsstemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowballsstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly") #example of Poter stemmer
```

```python
snowballsstemmer.stem("fairly"),snowballsstemmer.stem("sportingly")
```
