### OneHotEncoding with Top Categories

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
replace = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(replace, 'uncommon')).sample(5)
```
