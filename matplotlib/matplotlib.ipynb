```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# Lables title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
# Color(hex) line(width and style) and marker(size)
#we can give any color using hex code - search on google hex code and copy the hex code 
# plt.plot(batsman['index'],batsman['V Kohli'],color = 'green')
plt.plot(batsman['index'],batsman['V Kohli'],color = '#276CF5') #hex code
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black')
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
# linestyle 
plt.plot(batsman['index'],batsman['V Kohli'],color = '#276CF5',linestyle = 'dashed') 
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle = 'dashed')
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
# linestyle 
plt.plot(batsman['index'],batsman['V Kohli'],color = '#276CF5',linestyle = 'dotted') 
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle = 'dotted')
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
# linestyle  and width
plt.plot(batsman['index'],batsman['V Kohli'],color = '#276CF5',linestyle = 'dashdot',linewidth = 3) 
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'black',linestyle = 'dashdot',linewidth=2)
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
# Marker 
plt.plot(batsman['index'],batsman['V Kohli'],color = '#276CF5',linestyle = 'dashdot',linewidth = 3,marker = 'D',markersize=10) 
plt.plot(batsman['index'],batsman['RG Sharma'],color = '#F54927',linestyle = 'dashdot',linewidth=2,marker = '+',markersize=10)
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')
```

```python
#lable line name - Legend -> Location
plt.plot(batsman['index'],batsman['V Kohli'],color = '#276CF5',linestyle = 'dashdot',linewidth = 3,marker = 'D',markersize=8,label = 'Virat Kohli') 
plt.plot(batsman['index'],batsman['RG Sharma'],color = '#F54927',linestyle = 'dashdot',linewidth=2,marker = 'o',markersize=10,label = 'Rohit Sharma')
plt.title('Rohit Sharma vs Virat Kohli career Comparison')
plt.xlabel('Seasons')
plt.ylabel('Runs Scored')

plt.legend() #this is call to represent label
```

### Bar Chart
- Bivariate Analysis or Univariate Analysis
- Numerical vs Categorical 
- Use case - Aggregate Analysis of groups

```python
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color = 'red') #(x_axis , y_axis)
```

```python
# rotaion 
children = [10,20,40,10,30]
colors = ['red red','blue blue blue','green green green','yellow yellow yellow','pink pink pink']
plt.bar(colors,children,color = 'red') 
plt.xticks(rotation = 'vertical')
```

```python
#Horizontal bar chart - barh
plt.barh(colors,children,color = 'red') #optional - if we have more number of catagories use horizontal bar chart 
```

```python
df = pd.read_csv(r'D:\Learning Data Science\matplotlib\Dataset\batsman_season_record.csv')
df
#dispplay the bar chart for  2015 , define width 
plt.bar(df['batsman'],df['2015'],width=0.1)
```

```python
#display bar chat for 2015,2016,2017 side by side bar chart 
plt.bar(np.arange(df.shape[0])- 0.2,df['2015'],width=0.2,color = 'red')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='green')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='black')

plt.xticks(np.arange(df.shape[0]),df['batsman']) #plt.xticks('existing label','new label')

plt.show()
```

```python
# Stacked Bar Chart
df
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label ='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label = '2015') #we have to add all above 

plt.legend()
plt.show()
```

#### Pie Chart
- Univariate/Bivarite Analysis
- Categorical VS Numerical data
- use case - To find contribution on a standard scale

```python
data  = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)
plt.show()
```

```python
df = pd.read_csv(r'D:\Learning Data Science\matplotlib\Dataset\gayle-175.csv')
df
```

```python
#autopct = '%0.1f%%' -> Percentage 
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','brown','red'])
plt.show()
```

```python
# Explode and shadow 
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0,0,0,0,0.1],shadow=True)
plt.show()
```
