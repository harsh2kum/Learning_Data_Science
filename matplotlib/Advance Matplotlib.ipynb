### SubsPlots
- contain to axis - fig , axis 
- eg - fig , ax = plt.subplots()
- we can use them to plot 

```python
batters.head()
# 1 way to represent Scatterplot
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title("Batsman Data")
plt.xlabel('Average')
plt.ylabel('Strike_rate')
plt.show()  #also use to hide technical details
```

```python
# 2nd way to represent Scatterpot
# main thing is to use this ways,because it will help use to plot side by side graph 
fig , ax = plt.subplots(figsize = (15,6))
ax.scatter(batters['avg'],batters['strike_rate'],color = 'red',marker='+')
ax.set_title('Batsman Data')
ax.set_xlabel('Average')
ax.set_ylabel('Strik_rate')
fig.show()
```

```python
plt.subplots(nrows=2,ncols=1) # 2x1 = 2 graphs
plt.subplots(nrows=2,ncols=2) # 2x2 = 4 graphs 
plt.subplots(nrows=4,ncols=4) # 4x4 = 16 graphs
# we can plot different number of graphs
```

```python
fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True,figsize =(10,6)) # ax has 2 axis -> [<Axes: >, <Axes: >], sharex = True will share X axis with both the graph
ax[0].scatter(batters['avg'],batters['strike_rate'],color='red') #1st Axis 
ax[1].scatter(batters['avg'],batters['runs']) #2nd Axis

ax[0].set_title('Avg vs Strik Rate')
ax[0].set_ylabel('Strik Rate')

ax[1].set_title('Avg vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10)) 
ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='red') #1st Axis 
ax[0,1].scatter(batters['avg'],batters['runs']) #2nd Axis
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
```

```python
# Another ways to represent the graphs
fig = plt.figure()

ax1 = fig.add_subplot(211) #211 - 2 row,1 columns , 1st graph
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(212) # 212 - 2 rows,1 columns , 2nd graph
ax2.hist(batters['runs'])
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1) #2,2,1 - 2 row,2 columns , 1st graph
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2) # 2,2,2 - 2 rows,2 columns , 2nd graph
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3) # 2,2,3 - 2 rows,2 columns , 3rd graph
ax3.hist(batters['avg'])
```

### 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx , yy = np.meshgrid(x,y)
xx.shape
yy.shape
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig  = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig  = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap = 'viridis')
fig.colorbar(p)
```

### Contour Plots - convert 3D to 2D

```python
# Contour 
fig  = plt.figure(figsize=(12,8))

ax = plt.subplot()
p = ax.contour(xx,yy,z,cmap = 'viridis')
fig.colorbar(p)
```

```python
# Countourf
fig  = plt.figure(figsize=(12,8))

ax = plt.subplot()
p = ax.contourf(xx,yy,z,cmap = 'viridis')
fig.colorbar(p)
```
