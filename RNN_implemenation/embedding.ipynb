```python
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding
from tensorflow.keras.models import Sequential
```

```python
## Sentence 
sent = ['the glass of milk',
        'the glass of juice',
        'the cup of tea',
        'I am a good boy',
        'I am a good developer',
        'understand the meaning of words',
        'your videos are good',]
```

```python
sent
```

```python
""" 
Principal Component Analysis (PCA) for visualizing word vectors. It shows how similar words tend to cluster together in vector space, which is significant for applications like recommendation systems.
"""
```

```python
## Define the vocubulary size 
voc_size = 10000
```

```python
## One Hot Representation

""" One-Hot Representation: where each word in a sentence is converted into a vector according to its index in a predefined vocabulary size (set at 10,000 in this case). Examples of sentences are provided to illustrate how each word is represented as a vector based on these indices."""


one_hot_rep = [one_hot(words,voc_size) for words in sent] # List comprehension
one_hot_rep
```

```python
import numpy as np
import pandas as pd 
```

```python
""" 
Padding Sequences: To handle variance in sentence lengths, the instructor discusses the importance of padding sequences by adding zeros to ensure all sentences are of uniform length. This step is crucial for effective model training. 
"""
sent_length = 8
embedded_docs = pad_sequences(one_hot_rep, padding='pre', maxlen = sent_length)
print(embedded_docs)
# there the 2 'pre'-place zero before and 'post'- place after 
```

```python
sent
```

```python
# Feature Representation
dim = 10  # Dimensions
```

```python
from tensorflow.keras import Input

model = Sequential()
model.add(Input(shape=(sent_length,)))
model.add(Embedding(voc_size, dim))
model.compile(optimizer='adam', loss='mse')
model.summary()
```

```python
model.predict(embedded_docs)
```
