#### Reference Counting 
Reference Counting is the primary method Python uses to manage memory. 
Each object in Python maintains a count of references pointing to it. 
When the reference count drops to zero, the memory occupied by the object is deallocated.



```python
import sys

a = []
print(sys.getrefcount(a))
# Output- 2 (because one reference from 'a' and one from getrefcount())
print(sys.getrefcount(a))
```

```python
b = a
print(sys.getrefcount(b))
```

### Garbage Collection 
Python includes a cyclic garbage collector to handle reference cycles. Reference cycles occur when objects reference each other, preventing their rerference counts from reaching zero.

```python
import gc #garbage collector
# To enable garbage collector
gc.enable()
```

```python
# To disable garbage collector
gc.disable()
```

```python
# To manually trigger the garbage collection
gc.collect()
```

```python
# Get Garbage collection Stats 
print(gc.get_stats()) #it will Return a list of dictionaries containing per-generation statistics.
```

```python
# Get unreachable objects
print(gc.garbage)
```

#### Memory Management Best Practices
- Use Local Variables: Local Variables have a shorter lifespan and are sooner than global variables.
- Avoid Circular References: Circular references can lead to memory leaks if not properly managed.
- Use Generators: Generators produce items one at a time nd only keep ne item in memory at a time, making them memory efficient.
- Explicitly Delete Objects: Use memory profiling tools like tracemalloc and memory_profiler to identify memory leaks and optimize memory usage.


```python
# Handled Circular Reference
import gc

class MyObjects:
    def __init__(self,name):
        self.name = name
        print(f'Object {self.name} Created')

    def __del__(self):
        print(f'Object {self.name} deleted')

# Create Circular reference
obj1 = MyObjects("obj1")
obj2 = MyObjects("Obj2")

# In garbage Collection avoid circular Reference
obj1.ref = obj2
obj2.ref = obj1 


del obj1
del obj2

# Manually trigger the garbage collection
gc.collect()

```

```python
# Print collected objects 
print(f'Garbage Collected Objects: {gc.garbage}')
```

```python
## Generators for memeory Efficiency
# Generators allow you to produce items one at a time, using memory efficiently by only keeping one item in memory at a time.

def generate_numbers(n):
    for i in range(n):
        yield i

# using the generator
for num in generate_numbers(100000):
    print(num)
    if num > 10:
        break
```

```python
# Profiling Memory Usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()
    
    create_list()
    
    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')


    print('[Top 10]')
    for stat in top_stats[::]:
        print(stat)
```

```python
main()
```
